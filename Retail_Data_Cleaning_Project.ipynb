```python
import pandas as pd
import numpy as np

df = pd.read_excel("OnlineRetail.xlsx")

#print(df.head())

#print(df.shape)


print(df.isnull().sum())

print("duplicate rows", df.duplicated().sum())

print(df.describe())

missing_values = df.isnull().sum()

print(missing_values)

percen = ((df.isnull().sum())/len(df))*100

print(percen.astype(str) + "%")


# Step 4: Count unusual numerical values

print("Negative quantities:",
      (df["Quantity"] < 0).sum())

print("Zero quantities:",
      (df["Quantity"] == 0).sum())

print("Zero prices:",
      (df["UnitPrice"] == 0).sum())

print("Negative prices:",
      (df["UnitPrice"] < 0).sum())
```

```python

# Step 5: Create a summary of data quality issues

audit_summary = pd.DataFrame({
    "Issue": [
        "Total rows",
        "Total columns",
        "Missing CustomerID",
        "Missing Description",
        "Duplicate rows",
        "Negative quantities",
        "Zero prices",
        "Negative prices"
    ],
    "Count": [
        len(df),
        len(df.columns),
        df["CustomerID"].isnull().sum(),
        df["Description"].isnull().sum(),
        df.duplicated().sum(),
        (df["Quantity"] < 0).sum(),
        (df["UnitPrice"] == 0).sum(),
        (df["UnitPrice"] < 0).sum()
    ]
})

audit_summary

print(percen.astype(str) + "%")
```

```python

# Create a copy of the original dataset
df_clean = df.copy()

# Check the number of rows and columns
print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python

# Count duplicates before removal
duplicates_before = df_clean.duplicated().sum()

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Count duplicates after removal
duplicates_after = df_clean.duplicated().sum()

print("Duplicates before cleaning:", duplicates_before)
print("Duplicates after cleaning:", duplicates_after)
print("Shape after removing duplicates:", df_clean.shape)
```

```python

# Select rows where Description is missing
missing_description_rows = df_clean[
    df_clean["Description"].isnull()
]

print("Rows with missing Description:",
      len(missing_description_rows))

missing_description_rows.head(10)
```

```python

# Select rows where Description is missing
missing_desc = df_clean[
    df_clean["Description"].isnull()
]

# Count how many have zero prices
zero_price_missing_desc = (
    missing_desc["UnitPrice"] == 0
).sum()

# Count how many have non-zero prices
nonzero_price_missing_desc = (
    missing_desc["UnitPrice"] != 0
).sum()

print("Missing descriptions:", len(missing_desc))
print("With zero prices:", zero_price_missing_desc)
print("With non-zero prices:", nonzero_price_missing_desc)
```

```python

# Count missing descriptions with negative quantities
negative_quantity_missing_desc = (
    missing_desc["Quantity"] < 0
).sum()

# Count missing descriptions with positive quantities
positive_quantity_missing_desc = (
    missing_desc["Quantity"] > 0
).sum()

print("Negative quantities:", negative_quantity_missing_desc)
print("Positive quantities:", positive_quantity_missing_desc)
```

```python

# Check whether missing-description rows are cancellations

missing_desc = df_clean[
    df_clean["Description"].isnull()
]

cancelled_missing_desc = missing_desc[
    missing_desc["InvoiceNo"].astype(str).str.startswith("C")
]

print("Missing descriptions:", len(missing_desc))
print("Missing descriptions with cancelled invoices:",
      len(cancelled_missing_desc))

cancelled_missing_desc.head(10)
```

```python

# Identify cancelled invoices
cancelled_rows = df_clean[
    df_clean["InvoiceNo"].astype(str).str.startswith("C")
]

print("Cancelled transaction rows:", len(cancelled_rows))

cancelled_rows.head(10)
```

```python

# Inspect the quantities in cancelled transactions

print("Negative quantities:",
      (cancelled_rows["Quantity"] < 0).sum())

print("Zero quantities:",
      (cancelled_rows["Quantity"] == 0).sum())

print("Positive quantities:",
      (cancelled_rows["Quantity"] > 0).sum())
```

```python

# Create the cleaned master dataset
# Exact duplicate rows were already removed

df_master_clean = df_clean.copy()

print("Original rows:", len(df))
print("Rows after duplicate removal:", len(df_master_clean))
print("Remaining duplicates:",
      df_master_clean.duplicated().sum())
```

```python

# Step 1: Exclude cancelled invoices
sales_df = df_master_clean[
    ~df_master_clean["InvoiceNo"].astype(str).str.startswith("C")
].copy()

# Step 2: Remove records without product descriptions
sales_df = sales_df[
    sales_df["Description"].notna()
].copy()

# Step 3: Keep positive quantities
sales_df = sales_df[
    sales_df["Quantity"] > 0
].copy()

# Step 4: Keep positive unit prices
sales_df = sales_df[
    sales_df["UnitPrice"] > 0
].copy()

# Step 5: Check the result
print("Rows for product-sales analysis:", len(sales_df))
print("Columns:", len(sales_df.columns))

print("Missing descriptions:",
      sales_df["Description"].isnull().sum())

print("Non-positive quantities:",
      (sales_df["Quantity"] <= 0).sum())

print("Non-positive prices:",
      (sales_df["UnitPrice"] <= 0).sum())

print("Duplicate rows:", sales_df.duplicated().sum())
```

```python

# Create a before-and-after data quality report

quality_report = pd.DataFrame({
    "Metric": [
        "Total rows",
        "Duplicate rows",
        "Missing CustomerID",
        "Missing Description",
        "Negative quantities",
        "Zero prices",
        "Negative prices"
    ],

    "Original Dataset": [
        len(df),
        df.duplicated().sum(),
        df["CustomerID"].isnull().sum(),
        df["Description"].isnull().sum(),
        (df["Quantity"] < 0).sum(),
        (df["UnitPrice"] == 0).sum(),
        (df["UnitPrice"] < 0).sum()
    ],

    "Cleaned Master Dataset": [
        len(df_master_clean),
        df_master_clean.duplicated().sum(),
        df_master_clean["CustomerID"].isnull().sum(),
        df_master_clean["Description"].isnull().sum(),
        (df_master_clean["Quantity"] < 0).sum(),
        (df_master_clean["UnitPrice"] == 0).sum(),
        (df_master_clean["UnitPrice"] < 0).sum()
    ],

    "Product Sales Dataset": [
        len(sales_df),
        sales_df.duplicated().sum(),
        sales_df["CustomerID"].isnull().sum(),
        sales_df["Description"].isnull().sum(),
        (sales_df["Quantity"] < 0).sum(),
        (sales_df["UnitPrice"] == 0).sum(),
        (sales_df["UnitPrice"] < 0).sum()
    ]
})

quality_report
```

```python

# Create a sales value column

sales_df["SalesValue"] = (
    sales_df["Quantity"] * sales_df["UnitPrice"]
)

# Display sample results
sales_df[
    ["InvoiceNo", "Description", "Quantity",
     "UnitPrice", "SalesValue"]
].head(10)
```

```python

# Export the datasets and quality report

df_master_clean.to_csv(
    "retail_cleaned_master.csv", index=False
)

sales_df.to_csv(
    "retail_product_sales.csv", index=False
)

quality_report.to_csv(
    "data_quality_report.csv", index=False
)

print("Project files exported successfully.")
```
